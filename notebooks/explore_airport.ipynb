```python
from google.cloud import bigquery

client = bigquery.Client(project="prc-data-challenge-2025")


def q(sql):
    return client.query(sql).to_dataframe()


# preview 10 rows
q("""
SELECT *
FROM `prc-data-challenge-2025.prc2025.airports`
LIMIT 10
""")
```

Each line is an airport, with geographical coordonates, and an elevation (altitude) in ft -> to determine. And this column has some NA.

```python
# row count, distinct icao and null elevation
q("""
SELECT COUNT (*) as n_nows,
COUNT (DISTINCT icao) as n_airports,
COUNTIF(elevation IS NULL) as n_null_elevation
FROM `prc-data-challenge-2025.prc2025.airports`
""")
```

So one row = one unique airport, and 219 of them do not have an altitude information.

```python
# elevation quartiles
q("""
SELECT 
APPROX_QUANTILES(elevation, 4) as elevation_quartiles,
FROM `prc-data-challenge-2025.prc2025.airports`
""")
```

Half of the airports are under 529 ft, half above.
Min -1266 ft (Bar Yehuda, Dead Sea) and max 14472 ft (Daocheng Yading, China): both real, not errors.->
